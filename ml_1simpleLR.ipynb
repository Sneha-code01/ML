```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


sr1 = pd.read_csv("/content/subs.csv")
val = np.arange(0,365)
sr1['index'] = val
print(sr1)
x = sr1.iloc[:,1:]
y = sr1.iloc[:,0:1]
print(x)
print(y)
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state=2)

from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
print("test data of input:",x_test)
print("test data of output:",y_test)
n = int(input("enter the index of the row for which you want predicton of subscriber:"))
tpredict = print("the predicted subscriber be:",lr.predict(x_test.iloc[n].values.reshape(1,1)))
plt.scatter(sr1['index'],sr1['Subscribers gained'])
plt.plot(x_test,lr.predict(x_test),c='red')
m = lr.coef_
b = lr.intercept_
result = m*x+ b
print(result)
```
