```python
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

class SGDRegressor:

    def __init__(self, lr=0.01, epochs=100):
        self.lr = lr
        self.epochs = epochs
        self.coef_ = None
        self.intercept_ = None

    # Training function
    def fit(self, X, y):

        # Number of samples and features
        n_samples, n_features = X.shape

        # Initialize weights and bias
        self.coef_ = np.ones(n_features)
        self.intercept_ = 0

        # SGD loop
        for i in range(self.epochs):

            # Go through every sample
            for j in range(n_samples):

                # Randomly choose an index
                idx = np.random.randint(0, n_samples)

                # Prediction for that sample
                y_hat = np.dot(X[idx], self.coef_) + self.intercept_

                # Calculate derivatives
                slope_b = -2 * (y[idx] - y_hat)

                slope_m = -2 * (y[idx] - y_hat) * X[idx]

                # Update parameters
                self.intercept_ = self.intercept_ - self.lr * slope_b

                self.coef_ = self.coef_ - self.lr * slope_m

        print(self.intercept_, self.coef_)


    # Prediction function
    def predict(self, X):

        return np.dot(X, self.coef_) + self.intercept_

```
