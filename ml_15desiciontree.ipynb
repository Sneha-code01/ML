```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier

def analyzer(max_depth):

    # Load dataset
    data = pd.read_csv('/content/Social_Network_Ads (1).csv')

    # Features
    X = data.iloc[['Age','EstimatedSalary']].values   # Age, EstimatedSalary

    # Target
    y = data.iloc[:, -1].values    # Purchased

    # Train model
    clf = DecisionTreeClassifier(
        max_depth=max_depth,
        random_state=42
    )

    clf.fit(X, y)

    # Create mesh grid
    x_min = X[:, 0].min() - 1
    x_max = X[:, 0].max() + 1

    y_min = X[:, 1].min() - 1000
    y_max = X[:, 1].max() + 1000

    xx, yy = np.meshgrid(
        np.arange(x_min, x_max, 0.1),
        np.arange(y_min, y_max, 100)
    )

    # Predict for every point
    grid_points = np.c_[xx.ravel(), yy.ravel()]

    Z = clf.predict(grid_points)

    # Reshape for contour plot
    Z = Z.reshape(xx.shape)

    # Plot decision boundary
    plt.figure(figsize=(10,6))

    plt.contourf(xx, yy, Z, alpha=0.3)

    plt.scatter(
        X[:,0],
        X[:,1],
        c=y,
        edgecolors='k'
    )

    plt.xlabel('Age')
    plt.ylabel('Estimated Salary')
    plt.title(f'Decision Tree (max_depth={max_depth})')

    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier
```

```python
data = pd.read_csv("/content/Social_Network_Ads (1).csv")

print(data.head())

X = data[['Age','EstimatedSalary']].values
y = data['Purchased'].values

print(X.shape)
```

```python
def analyzer(max_depth):

    clf = DecisionTreeClassifier(
        max_depth=max_depth,
        random_state=42
    )

    clf.fit(X,y)

    x_min = X[:,0].min()-1
    x_max = X[:,0].max()+1

    y_min = X[:,1].min()-1000
    y_max = X[:,1].max()+1000

    xx,yy = np.meshgrid(
        np.arange(x_min,x_max,0.1),
        np.arange(y_min,y_max,100)
    )

    input_array = np.c_[xx.ravel(),yy.ravel()]

    labels = clf.predict(input_array)

    labels = labels.reshape(xx.shape)

    plt.figure(figsize=(10,6))

    plt.contourf(xx,yy,labels,alpha=0.3)

    plt.scatter(
        X[:,0],
        X[:,1],
        c=y,
        edgecolors='k'
    )

    plt.xlabel("Age")
    plt.ylabel("Estimated Salary")
    plt.title(f"Decision Tree (max_depth={max_depth})")

    plt.show()
```

```python
analyzer(4)
```
