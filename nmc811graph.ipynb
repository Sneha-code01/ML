```python
import matplotlib.pyplot as plt
import numpy as np

stages = [
    "Lithium Hydroxide",
    "Nickel Sulphate",
    "Cobalt Sulphate",
    "Manganese Sulphate",
    "Ammonium Hydroxide",
    "Cathode Precursor",
    "LiOH Transport",
    "Final NMC811"
]

ghg = [25.023,7.82,6.23,1.22,1.30,2.48,0.75,28.67]

# Beautiful color gradient
colors = plt.cm.plasma(np.linspace(0.2,0.9,len(stages)))

plt.figure(figsize=(12,7), dpi=300)

bars = plt.barh(stages, ghg,
                color=colors,
                edgecolor='black',
                linewidth=1)

plt.xlabel("GHG-100 Emission (kg CO$_2$-eq)",fontsize=13,fontweight='bold')
plt.ylabel("Production Stage",fontsize=13,fontweight='bold')
plt.title("Stage-wise Greenhouse Gas Emissions during NMC811 Production",
          fontsize=16,fontweight='bold')

plt.grid(axis='x',linestyle='--',alpha=0.4)

for bar in bars:
    width = bar.get_width()
    plt.text(width+0.3,
             bar.get_y()+bar.get_height()/2,
             f'{width:.2f}',
             fontsize=11,
             va='center',
             fontweight='bold')

plt.gca().spines[['top','right']].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

stages=[
"Lithium Hydroxide",
"Nickel Sulphate",
"Cobalt Sulphate",
"Manganese Sulphate",
"Ammonium Hydroxide",
"Final NMC811"
]

sox=[32.14,0.26,24.29,2.22,0.0184,0.35]

colors=plt.cm.Blues(np.linspace(0.35,0.95,len(stages)))

plt.figure(figsize=(11,6),dpi=300)

bars=plt.barh(stages,sox,color=colors,
              edgecolor='black')

plt.title("Stage-wise SO$_x$ Emissions",
          fontsize=16,
          fontweight='bold')

plt.xlabel("SO$_x$ Emission (g)",
           fontsize=13,
           fontweight='bold')

plt.grid(axis='x',alpha=0.35)

for i,v in enumerate(sox):
    plt.text(v+0.3,i,
             f"{v:.2f}",
             va='center',
             fontweight='bold')

plt.gca().spines[['top','right']].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

stages=[
"Lithium Hydroxide",
"Nickel Sulphate",
"Cobalt Sulphate",
"Ammonium Hydroxide",
"Cathode Precursor",
"Final NMC811"
]

voc=[4.54,2.46,1.65,2.55,0.108,6.10]

colors=plt.cm.Greens(np.linspace(0.35,0.95,len(stages)))

plt.figure(figsize=(11,6),dpi=300)

bars=plt.barh(stages,voc,
              color=colors,
              edgecolor='black')

plt.title("Stage-wise VOC Emissions",
          fontsize=16,
          fontweight='bold')

plt.xlabel("VOC Emission (g)",
           fontsize=13,
           fontweight='bold')

plt.grid(axis='x',alpha=0.3)

for i,v in enumerate(voc):
    plt.text(v+0.08,i,
             f"{v:.2f}",
             va='center',
             fontweight='bold')

plt.gca().spines[['top','right']].set_visible(False)

plt.tight_layout()
plt.show()
```
